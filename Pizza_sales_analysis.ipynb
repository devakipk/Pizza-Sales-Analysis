Project Name :- Pizza Sales Analysis

Objective :- This project uses python and it's libraries to analysis pizza sales data and finds insights from that which will helpful to take decisions about pizza sales.

Step 1 :- Loading library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/pizza_sales.csv')
```

```python
print("Fetching first 10 records :\n",df.head(10))
```

```python
print("Fetching last 10 records :\n",df.tail(10))
```

Step 2 :- Data exploration

```python
# print basic information about data

print("Basic info about data : \n")
df.info()

print("\nTotal no of rows :",df.shape[0])

print("\nTotal no of columns :", df.shape[1])

print("\nTotal no of records :",df.size)

print("\nDescribing numerical columns :\n",df.describe())
```

```python
# display column lists

print("Column List :\n",df.columns.tolist())
```

Step 3 :- Data Cleaning

```python
# checking null values

print("Counting null values :\n",df.isnull().sum())
```

```python
# checking for duplicates

print(" Checking Duplicates :\n",df.duplicated().sum())
```

Step 4 :- Data Evaluation

```python
print("Total Count of pizza by its category:\n")
df.groupby(['pizza_category','pizza_name'])['pizza_name'].value_counts()
```

```python

df.groupby("pizza_category").agg(Total_Revenue=("total_price", "sum"), Total_orders=("order_id", "nunique"), Total_pizza_sold=("quantity", "sum"))
```

```python

df.groupby("pizza_size").agg(Total_Revenue=("total_price", "sum"), Total_orders=("order_id", "nunique"), Total_pizza_sold=("quantity", "sum"))
```

```python
print("Top 5 best selling pizza :\n")
df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False).head(5)
```

```python
print("Least 5 selling pizza :\n")
df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=True).head(5)
```

```python
print("Top 5 pizza by its high revenue:\n")
df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).head(5)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['month'] = df['order_date'].dt.month_name()
df['date'] = df['order_date'].dt.date

df.groupby('month').agg(
    Total_Revenue=('total_price', 'sum'),
    Total_Orders=('order_id', 'nunique'),
    Total_Pizzas=('quantity', 'sum')
)
```

Step 5 :- Data Visualization

```python
# plotting multiple graphs

plt.figure(figsize=(10, 6))

# graph 1:quantity sold by category
plt.subplot(2,3,1)
X = df.groupby("pizza_category")["quantity"].sum()

plt.pie(X.values,labels= X.index,autopct='%1.1f%%')

plt.title("Quantity Sold By It's Category ")

# graph 2: total revenue by category
plt.subplot(2,3,2)

Y= df.groupby("pizza_category")["total_price"].sum()

plt.bar(x=Y.index,height=Y.values, color='Blue')

plt.title("Total Revenue By It's Category ")
plt.xlabel(" Pizza Category ")
plt.ylabel(" Total Revenue")

#graph 3:Total orders places by category
plt.subplot(2,3,3)
Z = df.groupby("pizza_category")["order_id"].nunique()

plt.bar(x=Z.index,height=Z.values,color='Green')

plt.title("Orders placed By It's Category ")
plt.xlabel(" Pizza Category ")
plt.ylabel(" Orders ")


#graph 4 : total quantity sold by pizza size
plt.subplot(2,3,4)

M = df.groupby("pizza_size")["quantity"].sum()

plt.pie(M.values,labels=M.index,autopct='%1.1f%%')

plt.title("Quantity Sold By It's pizza size ")

#graph 5: total revenue by pizza size
plt.subplot(2,3,5)

N = df.groupby("pizza_size")["total_price"].sum()

plt.bar(x=N.index,height=N.values,color='Blue')

plt.title("Total Revenue By It's pizza size ")
plt.xlabel(" Pizza Size ")
plt.ylabel(" Total revenue")

#graph 6 : orders placed by pizza size
plt.subplot(2,3,6)

P = df.groupby("pizza_size")["order_id"].nunique()

plt.bar(x=P.index,height=P.values,color='Green')

plt.title("Orders placed By It's Pizza size ")
plt.xlabel(" Pizza Size ")
plt.ylabel(" Orders")

plt.tight_layout()
plt.show()
```

```python
# scatter plot to show unit price v/s quantity

plt.figure(figsize=(10,8))

plt.scatter(x=df["quantity"],y=df["unit_price"])

plt.title("Quantity v/s Unit Price")
plt.xlabel("Quantity ")
plt.ylabel(" Unit Price")

plt.show()
```

```python
# plotting graph to show distribution of quantity
plt.figure(figsize=(10,8))

X = df["quantity"]

# Only pass the data column (X) to the histogram
plt.hist(X, bins=30, rwidth=0.8, color='skyblue', edgecolor='black')

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.show()
```

```python
# plotting graph to show distribution of total Price
plt.figure(figsize=(10,8))

X = df["total_price"]

# Only pass the data column (X) to the histogram
plt.hist(X, bins=10, rwidth=0.8, color='skyblue', edgecolor='black')

plt.title("Distribution of Total Price")
plt.xlabel("Total Price")
plt.ylabel("Frequency")


plt.show()
```

```python
# plotting graph to show top 5 best selling pizzas

plt.figure(figsize=(10,6))

X = df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).head(5)

plt.bar(x=X.index, height=X.values, color="Blue")

plt.title("Top 5 Best Selling Pizzas ")
plt.xlabel("Pizza Names")
plt.ylabel("Pizza Sale")
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# plotting graph to show least 5 selling pizzas

plt.figure(figsize=(10,6))

X = df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=True).head(5)

plt.bar(x=X.index, height=X.values, color="red")

plt.title("Least 5 Selling Pizzas ")
plt.xlabel("Pizza Names")
plt.ylabel("Pizza Sale")
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# plotting graph to show monthly revenue

monthly_revenue = df.groupby('month').agg(
    Total_Revenue=('total_price', 'sum')
).reset_index()


plt.figure(figsize=(8, 8))

plt.pie(
    monthly_revenue['Total_Revenue'],
    labels=monthly_revenue['month'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Monthly Revenue Distribution')
plt.show()
```

```python
#Plotting graph to show daily revenue

daily_revenue = df.groupby('date').agg(
Total_Revenue=('total_price', 'sum')
).reset_index()

plt.figure(figsize=(12, 6))

plt.plot(
daily_revenue['date'],
daily_revenue['Total_Revenue'],
marker='o',
linewidth=2
)

plt.title('Daily Revenue Trend')
plt.xlabel('Date')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df['time'] = pd.to_datetime(df['order_time'])

df['hour'] = df['time'].dt.hour

orders_by_hour = df.groupby('hour').agg(
    Total_Orders=('order_id', 'nunique')
).reset_index()

plt.figure(figsize=(10,5))

plt.plot(
    orders_by_hour['hour'],
    orders_by_hour['Total_Orders'],
    marker='o'
)

plt.title('Total Orders by Hour')
plt.xlabel('Hour')
plt.ylabel('Total Orders')
plt.grid()
plt.show()
```

```python
# Group by pizza category and size to calculate total quantity sold
Totals = df.groupby(['pizza_category', 'pizza_size']).agg(
    Total_Quantity=('quantity', 'sum')).reset_index()

plt.figure(figsize=(10, 6))

# Using a grouped barplot which is ideal for Category + Size visualization
sns.barplot(data=Totals, x='pizza_category', y='Total_Quantity', hue='pizza_size')

plt.title("Quantity Sold By Category And Size")
plt.xlabel("Category")
plt.ylabel("Quantity")

plt.tight_layout()
plt.show()
```

```python
# Calculate the correlation matrix for numerical columns
corr_matrix = df.corr(numeric_only=True)

# Plot the correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(corr_matrix, annot=True, cmap='Blues')

plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
